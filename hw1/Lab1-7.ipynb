10. Как определить, есть ли в данном двумерном массиве пустые столбцы?

````python
array = np.array([[1., np.nan, 0.], [2., np.nan, 0.]])
nan_columns = np.sum(array != array, axis=0) == array.shape[0]
zero_columns = np.sum(array == 0, axis=0) == array.shape[0]

for j in range(array.shape[1]):
    if nan_columns[j]:
        print("Столбец из NaN:", j)
    if zero_columns[j]:
        print("Нулевой столбец:", j)
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
df[["Systolic", "Diastolic"]] = df["Blood Pressure (s/d)"].str.split("/", expand=True).astype(int)
stress_q1, stress_q3 = df["Stress Levels"].quantile([0.25, 0.75])
low_stress = df[df["Stress Levels"] <= stress_q1]
high_stress = df[df["Stress Levels"] >= stress_q3]
pressure = pd.DataFrame({
    "Низкий стресс": low_stress[["Systolic", "Diastolic"]].mean(),
    "Высокий стресс": high_stress[["Systolic", "Diastolic"]].mean(),
}).round(2)
print(f"Пороги: ≤ {stress_q1:.2f} и ≥ {stress_q3:.2f}; по {len(low_stress)} человек")
print(pressure)
print("Вывод: среднее давление при высоком стрессе немного выше, но разница мала;")
print("по этому сравнению нельзя установить причинную связь.")
````

6. Имеет ли связь зрение и способность слышать звуки на высоких частотах? Ответьте на вопрос - какая зависимость между ними и предположите почему зависимость именно такая?

````python
vision = df["Vision Sharpness"]
hearing = df["Hearing Ability (dB)"]
age = df["Age (years)"]
rho = vision.corr(hearing, method="spearman")
vision_residual = vision - np.polyval(np.polyfit(age, vision, 1), age)
hearing_residual = hearing - np.polyval(np.polyfit(age, hearing, 1), age)
partial_r = np.corrcoef(vision_residual, hearing_residual)[0, 1]
fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(data=df, x="Vision Sharpness", y="Hearing Ability (dB)",
                hue="Age (years)", palette="viridis", legend="brief",
                s=22, alpha=0.55, edgecolor=None, ax=ax)
ax.legend(title="Возраст, лет", bbox_to_anchor=(1.02, 1), loc="upper left")
ax.set(xlabel="Чёткость зрения", ylabel="Показатель слуха, дБ",
       title=f"Зрение и слух  ·  ρ = {rho:.2f}")
fig.tight_layout()
plt.show()
print("У людей с менее чётким зрением показатель слуха в дБ обычно выше.")
print("По цвету точек видно, что это чаще люди старшего возраста.")
print(f"После поправки на возраст связь почти пропадает (r = {partial_r:.2f}).")
print("Похоже, оба показателя связаны с возрастом; один не обязательно влияет на другой.")
````
